- **Granularidad:** con qué nivel de detalle se registra la unidad.

Sin estas ideas podemos duplicar conteos o comparar registros diferentes.

## Señales visibles de calidad

Una señal abre una pregunta; no autoriza una corrección inmediata.

- `isna` identifica valores faltantes.
- `duplicated` identifica filas idénticas.
- `value_counts` muestra códigos y frecuencias.

Primero preguntamos qué significa la señal y qué información permitiría interpretarla.

### Ejemplo: contar señales sin modificar

```python
faltantes = datos.isna().sum().sort_values(ascending=False)
print(faltantes)
print(f"Filas exactamente repetidas: {datos.duplicated().sum()}")
print("\nCódigos de Type:")
print(datos["Type"].value_counts(dropna=False))
print("\nCódigos de Method:")
print(datos["Method"].value_counts(dropna=False))
```
