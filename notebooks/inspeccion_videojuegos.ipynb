```python
import pandas as pd

# Dataset principal
df = pd.read_csv('vgsales.csv')

# Fuente alternativa
df_alt = pd.read_csv('videoGameSalesWithRatings.csv')

print("Datasets cargados correctamente.")
print(f"Dataset principal -> Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
print(f"Fuente alternativa -> Filas: {df_alt.shape[0]}, Columnas: {df_alt.shape[1]}")
```

```python
# Mostrar las primeras 10 filas del dataset principal
df.head(10)
```

```python
# Información general: tipos de datos y valores no nulos
df.info()
```

```python
# Lista de columnas del dataset principal
print("Columnas del dataset vgsales.csv:")
for col in df.columns:
    print(f"- {col}")
```

```python
# Conteo de valores faltantes por columna
print("Valores faltantes por columna (dataset vgsales.csv):")
print(df.isnull().sum())
print("\nPorcentaje de faltantes:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
# Estadísticas descriptivas de columnas numéricas
df.describe()
```

```python
# Conteo de filas duplicadas
duplicados = df.duplicated().sum()
print(f"Filas duplicadas (dataset vgsales.csv): {duplicados}")
```

```python
# Valores únicos en columnas categóricas
print("Plataformas únicas:", df['Platform'].nunique())
print("Géneros únicos:", df['Genre'].nunique())
print("Publishers únicos:", df['Publisher'].nunique())
print("\nGéneros disponibles:")
print(df['Genre'].value_counts())
```

```python
# Rango de años cubiertos
print(f"Año mínimo: {df['Year'].min()}")
print(f"Año máximo: {df['Year'].max()}")
print(f"Registros sin año: {df['Year'].isnull().sum()}")
```

```python
# Información de la fuente alternativa
print("=== videoGameSalesWithRatings.csv ===")
print(f"Filas: {df_alt.shape[0]}, Columnas: {df_alt.shape[1]}")
print("\nColumnas:")
for col in df_alt.columns:
    print(f"- {col}")
print("\nValores faltantes:")
print(df_alt.isnull().sum())
```
